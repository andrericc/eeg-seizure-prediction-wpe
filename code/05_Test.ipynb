```python
from scipy.stats import wilcoxon
import numpy as np
import pandas as pd
```

```python
patients = ['chb01','chb03','chb04','chb05','chb06','chb07','chb08','chb09','chb10',
            'chb12','chb13','chb14','chb15','chb16','chb17','chb18','chb20','chb21',
            'chb22','chb23']

# Config A
auc_A = [0.688, 0.121, 0.660, 0.502, 0.638, 0.651, 0.129, 0.410, 0.685, 0.098,
         0.742, 0.443, 0.484, 0.835, 0.657, 0.410, 0.841, 0.848, 0.621, 0.573]

# Config B
auc_B = [0.893, 0.774, 0.631, 0.802, 0.614, 0.895, 0.779, 0.520, 0.489, 0.313,
         0.740, 0.572, 0.522, 0.805, 0.802, 0.715, 0.889, 0.673, 0.634, 0.482]

# Config C
auc_C = [0.651, 0.831, 0.503, 0.776, 0.684, 0.844, 0.885, 0.536, 0.581, 0.890,
         0.713, 0.566, 0.631, 0.922, 0.888, 0.168, 0.964, 0.748, 0.576, 0.490]

# Config D
auc_D = [0.65, 0.83, 0.50, 0.78, 0.68, 0.84, 0.89, 0.54, 0.58, 0.89,
         0.71, 0.57, 0.63, 0.92, 0.89, 0.17, 0.96, 0.75, 0.58, 0.49]

configs = {'A': np.array(auc_A), 'B': np.array(auc_B), 'C': np.array(auc_C), 'D': np.array(auc_D)}

comparisons = [('A', 'B'), ('B', 'C'), ('C', 'D'), ('A', 'D')]

results = []
for c1, c2 in comparisons:
    stat, p = wilcoxon(configs[c2], configs[c1])
    mean_diff = np.mean(configs[c2] - configs[c1])
    results.append({
        'Comparison': f'{c1} vs {c2}',
        'Mean AUC diff': round(mean_diff, 4),
        'Wilcoxon statistic': round(stat, 2),
        'p-value': round(p, 5)
    })

df_results = pd.DataFrame(results)
print(df_results.to_string(index=False))
```
